# A/B testing in digital data analytics

###A/B testing is a method used to compare two different versions in order to determine which one produces better results.It is most commonly used in the fields of marketing, product analytics, UX, and conversion optimization.

#### There are two variants in an A/B test:
#### A -> Existing version
#### B -> New version

#### Main objective:
#### Determine which version generates a higher conversion rate.

#### Real Business Example

#### Variant A -> Blue button
#### Variant B -> Green button

#### The company analyzes:
#### Which button generates more purchases?

# Hypotheses

# Null Hypothesis (H0):
# There is no difference between A and B.

# Alternative Hypothesis (H1):
# There is a difference between A and B.

```python
A qrupu → 1000 user → 120 conversion
B qrupu → 1000 user → 150 conversion
```

```python
#import library
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
```

```python
# conversions
success = np.array([120, 150])

# total users
samples = np.array([1000, 1000])
```

```python
z_stat, p_value = proportions_ztest(success, samples)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
```

#If p value < 0.05, there is significant difference
