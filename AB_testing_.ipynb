# Rəqəmsal data analitikasında A/B testləmə

###A/B testing iki fərqli versiyanı müqayisə edərək hansının daha yaxşı nəticə verdiyini müəyyən etmək üçün istifadə olunan bir üsuldur. Bu metod ən çox marketinq, product analytics, UX və conversion optimization sahələrində tətbiq olunur.

#### A/B testdə iki variant mövcuddur:
#### A -> Mövcud versiya
#### B -> Yeni versiya

#### Əsas məqsəd:
#### Hansı versiyanın daha yüksək conversion rate yaratdığını müəyyən etmək.

#### Real biznes nümunəsi

#### Variant A -> Blue button
#### Variant B -> Green button

#### Şirkət analiz edir:
#### Hansı düymə daha çox purchase yaradır?

# Hipotezlər

# Sıfır Hipotezi (H0)::
# A və B arasında heç bir fərq yoxdur.

# Alternativ Hipotez (H1):
# A və B arasında fərq vardır.

A qrupu → 1000 user → 120 conversion
B qrupu → 1000 user → 150 conversion

````python
#import library
import numpy as np
from statsmodels.stats.proportion import proportions_ztest
````

````python
# conversions
success = np.array([120, 150])

# total users
samples = np.array([1000, 1000])
````

````python
z_stat, p_value = proportions_ztest(success, samples)

print("Z-statistic:", z_stat)
print("P-value:", p_value)
````

#Əgər p-value < 0.05 olarsa, statistik olaraq əhəmiyyətli fərq mövcuddur.

Chi-Square Test (A/B Testing) -- ki kategorik dəyişən arasında əlaqə olub-olmadığını yoxlamaq üçün istifadə olunur.

````python
import pandas as pd

data = {
    "group": ["A"] * 1000 + ["B"] * 1000,
    "converted": (
        [1] * 80 + [0] * 920 +
        [1] * 120 + [0] * 880
    )
}

df = pd.DataFrame(data)

df.head()
````

````python
table = pd.crosstab(df["group"], df["converted"])

print(table)
````

````python
from scipy.stats import chi2_contingency
````

````python
chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
````



Gender → Qadın / Kişi
Accident → Edib / Etməyib

Məqsəd:
“Gender ilə qəza etmə arasında əlaqə varmı?”

Null Hypothesis (H0)

Gender ilə qəza etmə arasında əlaqə yoxdur.

Alternative Hypothesis (H1)

Gender ilə qəza etmə arasında əlaqə var.

````python
import pandas as pd

data = {
    "Gender": (
        ["Female"] * 100 +
        ["Male"] * 100
    ),

    "Accident": (
        ["Yes"] * 20 + ["No"] * 80 +
        ["Yes"] * 40 + ["No"] * 60
    )
}

df = pd.DataFrame(data)

print(df.head())
````

````python
df.info()
````

````python
table = pd.crosstab(df["Gender"], df["Accident"])

print(table)
````

20 qadın qəza edib
40 kişi qəza edib

İndi sual:
Bu fərq təsadüfidirmi?

Yoxsa gender ilə əlaqəlidir?

````python
from scipy.stats import chi2_contingency
````

````python
chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square Statistic:", chi2)
print("P-value:", p)
print("Degrees of Freedom:", dof)
````

````python
if p < 0.05:
    print("There is a significant relationship")
else:
    print("There is no significant relationship")
````

Business Problem

Bir e-commerce şirkəti iki fərqli endirim kampaniyasını test edir.

Variant A

“10% Discount”

Variant B

“Free Shipping”

Şirkət bilmək istəyir:

“Hansı kampaniya daha çox alış yaradır?”

H0

Kampaniya ilə alış arasında fərq yoxdur.

H1

Kampaniya ilə alış arasında fərq var.

````python
import pandas as pd

data = {
    "Campaign": (
        ["A"] * 500 +
        ["B"] * 500
    ),

    "Purchase": (
        ["Yes"] * 45 + ["No"] * 455 +
        ["Yes"] * 80 + ["No"] * 420
    )
}

df = pd.DataFrame(data)

print(df.head())
````

````python
table = pd.crosstab(df["Campaign"], df["Purchase"])

print(table)
````

````python
conversion_rate = df.groupby("Campaign")["Purchase"] \
    .apply(lambda x: (x == "Yes").mean() * 100)

print(conversion_rate)
````

````python
from scipy.stats import chi2_contingency

chi2, p, dof, expected = chi2_contingency(table)

print("Chi-Square:", chi2)
print("P-value:", p)
````

Business Problem

Bir delivery şirkəti iki fərqli marşrut sistemini test edir.

Variant A

Köhnə route system

Variant B

Yeni AI-based route system

Şirkət bilmək istəyir:

“Yeni sistem delivery vaxtını azaldırmı?”

H0

İki sistem arasında orta delivery vaxtında fərq yoxdur.

H1

Orta delivery vaxtında fərq var.

````python
import numpy as np

# Köhnə sistem
old_system = np.array([
    42, 39, 45, 50, 44,
    41, 47, 46, 43, 48
])

# Yeni sistem
new_system = np.array([
    35, 37, 36, 40, 38,
    34, 39, 36, 37, 35
])
````

````python
#Ortalamalari muayise et
print("Old system mean:", old_system.mean())
print("New system mean:", new_system.mean())
````

````python
from scipy.stats import ttest_ind
````

````python
#Test icra et
````

````python
t_stat, p = ttest_ind(old_system, new_system)

print("T-Statistic:", t_stat)
print("P-value:", p)
````

````python
if p < 0.05:
    print("Difference is statistically significant")
else:
    print("No significant difference")
````
